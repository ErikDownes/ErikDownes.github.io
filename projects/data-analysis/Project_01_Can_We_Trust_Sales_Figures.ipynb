# Project 01 | Can we trust these sales figures?

**Question:** A shop manager receives a sales export. Before reporting revenue, how can we establish whether the data is reliable? Work through evidence → cleaning → analysis → findings. Works in Colab and local Jupyter.

## Explore the evidence
Each row should represent a sale. Look for problems before calculating totals.

```python
import pandas as pd
from io import StringIO
raw = """sale_id,date,product,quantity,unit_price
101,2026-09-01,Notebook,3,4.50
102,2026-09-01,Pen,10,1.20
103,2026-09-02,Folder,2,3.00
104,2026-09-02,Notebook,1,4.50
105,2026-09-03,Pen,,1.20
106,2026-09-03,Folder,4,3.00
106,2026-09-03,Folder,4,3.00
107,2026-09-04,Notebook,2,4.50
108,2026-09-04,Pen,5,1.20
"""
sales = pd.read_csv(StringIO(raw))
display(sales)
```

## Check data quality
A blank quantity is *unknown*, not necessarily zero. A repeated sale ID needs investigation.

```python
print("Missing values:", sales.isna().sum().to_dict())
print("Repeated IDs:", sales.duplicated(subset="sale_id").sum())
```

## Make a defensible decision
For this exercise, retain the first occurrence of a sale ID and exclude transactions with unknown quantities from **confirmed** revenue. Real-world disputes require source checks.

```python
unique = sales.drop_duplicates(subset="sale_id", keep="first")
confirmed = unique.dropna(subset=["quantity"]).copy()
confirmed["revenue"] = confirmed["quantity"] * confirmed["unit_price"]
print(f"Confirmed revenue: €{confirmed['revenue'].sum():.2f}")
print("Excluded incomplete sales:", len(unique)-len(confirmed))
display(confirmed)
```

## Investigate the pattern
Which product generated most *confirmed* revenue? Does revenue measure the same thing as units sold?

```python
summary = confirmed.groupby("product", as_index=False).agg(units=("quantity","sum"),revenue=("revenue","sum"))
display(summary)
ax = summary.plot.bar(x="product", y="revenue", legend=False, rot=0, title="Confirmed sales revenue")
ax.set_ylabel("Euro (€)")
```
